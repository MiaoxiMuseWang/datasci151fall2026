**Question (D)** Find the maximum element of a list with missing data

<font size = "3">

Here is some code that uses a technique called a **running maximum** to find the largest element in a list:

````python
list_of_numbers = [25, 0, -1, -60, 30, 11, 55, 51, 11, 1, 5, 30]

current_max = float("-inf")     # start maximum at negative infinity

for number in list_of_numbers:
    if number > current_max:    # is this number bigger than the current maximum?
        current_max = number    # If so, update the value of current maximum

print("Largest number in list:", current_max)
````

````python
# write the function
num_list = []

current_max = float("-inf") 

def find_max(num_list):
    for number in num_list:
        if number > current_max:
            current_max = number
        if number == "missing":
            continue
    return find_max

````

**Question (F)** Computing square roots

<font size = "3">

How do computers actually compute square roots? A proper answer to this question would be too much for a DATASCI-151 quiz. Instead, you will code up a simplified version of **the Bakhshali method**, which is described in an ancient manuscript discovered in present day Pakistan (in the village of Bakhshali).

If $S > 0$ is a real number, and we want to compute an approximation to $\sqrt{S}$, we can do the following calculations:

- $x_0 = \dfrac{S}{2}$

- $a = \dfrac{S - x_0^2}{2x_0}$

- $x_1 = x_0 + a$

- $x_2 = x_1 - \dfrac{a^2}{2x_1}$

Then $x_1$ is an approximation to $\sqrt{S}$, and $x_2$ is (usually) an even better approximation.

Write a function that takes in a single argument `S`, which corresponds to a positive floating point number. Your function should have **two** return values, corresponding to $x_1$ and $x_2$, defined as in the equations above.

**Note:** I'm asking you to implement a *simplified* version of the Bakhshali method, so the approximations can be pretty bad for certain values of $S$. But if you test your function on values in the range $3 \leq S \leq 5$, you should see that $x_2$ is a pretty good approximation to `np.sqrt(S)`.


````python
# write your function here
import numpy as np

def fun(S):
    x_0 = S/2
    a = (S - (x_0 **2)) / (2 * x_0)
    x_1 = x_0 + a
    x_2 = x_1 - ((a**2) / (2*x_1))
    return x_1, x_2

print(fun(np.sqrt(3)))
````
